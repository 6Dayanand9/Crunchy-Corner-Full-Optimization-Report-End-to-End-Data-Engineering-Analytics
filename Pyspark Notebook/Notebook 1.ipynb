```python
# Welcome to your new notebook
# Type here in the cell editor to add code!
import pyspark.pandas as pd
```

```python
df = pd.read_csv("Files/Actual_uncleaned")
```

```python
df.head()
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df.fillna(0,inplace=True)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
      .str.replace("&", "and")
      .str.replace("-", "_")
      .str.replace("/", "_")
      .str.replace("__", "_")
)
```

```python
df.columns
```

```python
clean_df = df.to_spark()
```

```python
clean_df.write \
.format("delta") \
.mode("overwrite") \
.option("overwriteSchema", "true")\
.saveAsTable("silver_actual")
```
